```python
import random

def calculate_conflicts(state):
    conflicts = 0
    n = len(state)
    for i in range(n):
        for j in range(i + 1, n):
            if state[i] == state[j]:
                conflicts += 1
            elif abs(state[i] - state[j]) == abs(i - j):
                conflicts += 1
    return conflicts

def get_best_neighbor(state):
    n = len(state)
    current_conflicts = calculate_conflicts(state)
    best_neighbors = []
    min_conflicts = current_conflicts

    for col in range(n):
        original_row = state[col]
        for row in range(n):
            if row != original_row:
                neighbor = list(state)
                neighbor[col] = row
                neighbor_conflicts = calculate_conflicts(neighbor)

                if neighbor_conflicts < min_conflicts:
                    min_conflicts = neighbor_conflicts
                    best_neighbors = [neighbor]
                elif neighbor_conflicts == min_conflicts:
                    best_neighbors.append(neighbor)

    if best_neighbors:
        return min_conflicts, random.choice(best_neighbors)
    return current_conflicts, state

def hill_climbing_n_queens(n, max_restarts=1000):
    for restart in range(max_restarts):
        current_state = [random.randint(0, n - 1) for _ in range(n)]
        current_conflicts = calculate_conflicts(current_state)

        while True:
            if current_conflicts == 0:
                return current_state, restart

            neighbor_conflicts, neighbor_state = get_best_neighbor(current_state)

            if neighbor_conflicts >= current_conflicts:
                break

            current_state = neighbor_state
            current_conflicts = neighbor_conflicts

    return None, max_restarts

def print_board(state):
    if not state:
        print("No solution found.")
        return
    n = len(state)
    for row in range(n):
        line = ""
        for col in range(n):
            if state[col] == row:
                line += " Q "
            else:
                line += " . "
        print(line)

if __name__ == "__main__":
    print("Enter n: ")
    N=int(input())
    print(f"Solving {N}-Queens using Hill Climbing with Random Restarts...")
    solution, restarts_used = hill_climbing_n_queens(N)
    if solution:
        print(f"\nSolution found after {restarts_used} random restarts:\n")
        print_board(solution)
        print(f"\nState Array Representation: {solution}")
    else:
        print("\nFailed to find a solution within the maximum restart limit.")

```
