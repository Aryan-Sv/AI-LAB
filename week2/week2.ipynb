```python
import heapq

class PuzzleNode:
    def __init__(self, board, parent=None, move="", g=0, goal=None):
        self.board = board
        self.parent = parent
        self.move = move
        self.g = g
        self.goal = goal if goal else (1, 2, 3, 4, 5, 6, 7, 8, 0)
        self.h = self.calculate_manhattan_distance()
        self.f = self.g + self.h

    def calculate_manhattan_distance(self):
        distance = 0
        goal_positions = {}
        for index, val in enumerate(self.goal):
            if val != 0:
                goal_positions[val] = (index // 3, index % 3)

        for i, val in enumerate(self.board):
            if val != 0:
                current_row, current_col = i // 3, i % 3
                goal_row, goal_col = goal_positions[val]
                distance += abs(current_row - goal_row) + abs(current_col - goal_col)
        return distance

    def get_neighbors(self):
        neighbors = []
        blank_index = self.board.index(0)
        row, col = blank_index // 3, blank_index % 3

        moves = [
            (-1, 0, "Up"),
            (1, 0, "Down"),
            (0, -1, "Left"),
            (0, 1, "Right")
        ]

        for dr, dc, move_name in moves:
            new_row, new_col = row + dr, col + dc
            if 0 <= new_row < 3 and 0 <= new_col < 3:
                new_blank_index = new_row * 3 + new_col
                new_board = list(self.board)
                new_board[blank_index], new_board[new_blank_index] = new_board[new_blank_index], new_board[blank_index]
                neighbors.append(PuzzleNode(tuple(new_board), self, move_name, self.g + 1, self.goal))

        return neighbors

    def __lt__(self, other):
        return self.f < other.f

def Solve8Puzzle(start, goal):
    def is_solvable(start_board, goal_board):
        def count_inversions(board):
            inversions = 0
            flat_board = [x for x in board if x != 0]
            for i in range(len(flat_board)):
                for j in range(i + 1, len(flat_board)):
                    if flat_board[i] > flat_board[j]:
                        inversions += 1
            return inversions

        return count_inversions(start_board) % 2 == count_inversions(goal_board) % 2

    if not is_solvable(start, goal):
        return "Unsolvable layout"

    start_node = PuzzleNode(start, goal=goal)
    OPEN = [start_node]
    CLOSED = set()

    while OPEN:
        current = heapq.heappop(OPEN)

        if current.board == goal:
            path = []
            node = current
            while node:
                path.append((node.move, node.board))
                node = node.parent
            return path[::-1]

        CLOSED.add(current.board)

        for neighbor in current.get_neighbors():
            if neighbor.board in CLOSED:
                continue

            heapq.heappush(OPEN, neighbor)

    return "No Solution"

def print_board(board):
    for i in range(0, 9, 3):
        print(f"{board[i]} {board[i+1]} {board[i+2]}")
    print()

start_state = (2, 8, 3, 1, 6, 4, 7, 0, 5)
goal_state = (1, 2, 3, 8, 0, 4, 7, 6, 5)

print("Start State:")
print_board(start_state)

result = Solve8Puzzle(start_state, goal_state)

if isinstance(result, str):
    print(result)
else:
    print(f"Solved in {len(result) - 1} steps:\n")
    for step, (move, board) in enumerate(result):
        if step == 0:
            print("Step 0: Start Configuration")
        else:
            print(f"Step {step}: Slide blank '{move}'")
        print_board(board)
```

```python
graph = {
    'A': ['B', 'C'],
    'B': ['D', 'E'],
    'C': ['F', 'G'],
    'D': ['H'],
    'E': ['I'],
    'F': [],
    'G': ['J'],
    'H': [],
    'I': [],
    'J': []
}

def DLS(node, goal, depth, traversal):
    traversal.append(node)

    if node == goal:
        return "Found"

    if depth == 0:
        return "Not Found"

    for neighbor in graph.get(node, []):
        result = DLS(neighbor, goal, depth - 1, traversal)

        if result == "Found":
            return "Found"

    return "Not Found"

def IterativeDeepening(start, goal):
    depth = 0

    while True:
        traversal = []
        result = DLS(start, goal, depth, traversal)

        print(f"Depth={depth}    {' '.join(traversal)}")

        if result == "Found":
            print("\nResult found")
            print(f"Path= {' -> '.join(traversal)}")
            break

        depth += 1

IterativeDeepening('A', 'G')
```
