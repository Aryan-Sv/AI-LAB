```python
RoomA,RoomB='a','b'
rooms=[1,1]
x=RoomA
while(True):
  if(rooms[0]==rooms[1]==0):
    print("All rooms are clean")
    break
  elif(x=='a' and rooms[0]==1):
    print("Room A is dirty")
    print("Cleaning Room A")
    rooms[0]=0
  elif(x=='a' and rooms[0]==0):
    print("Moving to Room B")
    x=RoomB
  elif(x=='b' and rooms[1]==0):
    print("Room B is clean")
    print("Moving to Room A")
    x=RoomA
  elif(x=='b' and rooms[1]==1):
    print("Room B is dirty")
    print("Cleaning Room B")
    rooms[1]=0
```

```python
1
board = [
    [" ", " ", " "],
    [" ", " ", " "],
    [" ", " ", " "]
]
user = "X"
bot = "O"
def display_board():
    print()
    for i in range(3):
        print(" | ".join(board[i]))
        if i < 2:
            print("--+---+--")
    print()

def is_winning(player):
    for i in range(3):
        if (board[i][0] == player and
            board[i][1] == player and
            board[i][2] == player):
            return True
    for i in range(3):
        if (board[0][i] == player and
            board[1][i] == player and
            board[2][i] == player):
            return True
    if (board[0][0] == player and
        board[1][1] == player and
        board[2][2] == player):
        return True
    if (board[0][2] == player and
        board[1][1] == player and
        board[2][0] == player):
        return True
    return False

def is_board_full():

    for i in range(3):
        for j in range(3):
            if board[i][j] == " ":
                return False
    return True

def bot_move():
    for i in range(3):
        for j in range(3):
            if board[i][j] == " ":
                board[i][j] = user
                if is_winning(user):
                    board[i][j] = bot
                    return
                board[i][j] = " "
    for i in range(3):
        for j in range(3):
            if board[i][j] == " ":
                board[i][j] = bot
                if is_winning(bot):
                    return
                board[i][j] = " "
    for i in range(3):
        for j in range(3):
            if board[i][j] == " ":
                board[i][j] = bot
                return

while True:
    display_board()
    print("Your turn!")
    row = int(input("Enter row (1-3): ")) - 1
    col = int(input("Enter column (1-3): ")) - 1
    if row < 0 or row > 2 or col < 0 or col > 2:
        print("Invalid position!")
        continue
    if board[row][col] != " ":
        print("Position already occupied!")
        continue
    board[row][col] = user
    if is_winning(user):
        display_board()
        print("You Win!")
        break
    if is_board_full():
        display_board()
        print("Draw!")
        break
    print("Bot's turn...")
    bot_move()
    if is_winning(bot):
        display_board()
        print("Bot Wins!")
        break
    if is_board_full():
        display_board()
        print("Draw!")
        break
```
